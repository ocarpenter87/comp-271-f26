# Induction

In 1889, Italian mathematician Giuseppe Peano formalized arithmetic presenting several axioms about natural numbers. One of them introduced *Mathematical Induction*.

In simple terms, induction allows us to prove certain properties that involve natural numbers.

This is best demonstrated with an easy example.

We are asked to prove:

$$
1+2+\ldots +n = \dfrac{n(n+1)}{2}
$$

This can be done in three steps, based on Peano's axioms.

**First,** we test the property for a specific value of $n$. Most often we try $n=1$ but nothing stops us from trying any natural number, say $n=3$:

\begin{align*}
1+2+3 &= 6 
 = \dfrac{12}{2} 
 = \dfrac{3\times 4}{2}
 = \dfrac{3\times (3+1)}{2}
\end{align*}

So the property holds for $n=3$.

If the property holds for a specific value of $n$ as shown above, we move to the **second** step and we assume that the property holds for some arbitrary value $n=k$. In other words, we hold true that

$$
1+2+\ldots +k = \dfrac{k(k+1)}{2}
$$

This second step is usually the most confusing one. We essentially rewrite the property we wish to prove, but instead of $n$ we use $k$ and we *assume* its true? Well, if we assume it's true why bother proving it?

Because the essence of the proof is to demonstrate that the property holds for the next number up. That's the final step in the process.

The **final** step is to show that the property holds the next number up from $k$, i.e., $k+1$. In other words, what does the sum ${\color{blue}1+2+\ldots +k } + {\color{red}(k+1)}$ add up to?

From the second step above, we know that the blue part adds up to $k(k+1)/2$. Using that information, we can write

\begin{align*}
{\color{blue}1+2+\ldots +k } + {\color{red}(k+1)} & = \color{blue}\dfrac{k(k+1)}{2} + {\color{red}(k+1)} \\ \\
  &= \dfrac{{\color{blue}k(k+1)}+{\color{red}2(k+1)}}{2} \\ \\ 
  &= \dfrac{(k+1)({\color{blue}k}+{\color{red}2})}{2}
\end{align*}

Setting $z=k+1$, we just showed that $1+2+\ldots z = z(z+1)/2$ proving the property.

# Recursion is induction in disguise

Look back at what we just did. Induction has two moving parts, and recursion has the *same* two parts. That's not a coincidence: a recursive function is basically an induction proof that you can run.

| Induction | Recursion |
|---|---|
| **Base case:** show the property holds for a starting value | **Base case:** the input is small enough that we answer directly |
| **Inductive step:** assume it holds for $k$, then show it holds for $k+1$ | **Recursive case:** assume the function works on a smaller input, then use that answer to solve the current one |

Here's the sum we just proved, written as a function:

```python
def sum_to(n):
    if n == 0:
        result = 0                    # base case
    else:
        result = sum_to(n - 1) + n    # recursive case
    return result
```

Now read the code next to the proof:

- *"Test a specific value."* That's `if n == 0: result = 0`. Nothing to add up, so no work to do. It's our starting point, just like $n=1$ or $n=3$ was above.
- *"Assume the property holds for $k$."* That's the call `sum_to(n - 1)`. We don't peek inside it or worry about how it works. We just trust that it hands back $1+2+\ldots +(n-1)$. That trust is exactly the inductive hypothesis.
- *"Show it holds for $k+1$."* That's `... + n`. Take the answer we trusted (the blue part) and tack on the next number (the red part). It's the same move as $\color{blue}\dfrac{k(k+1)}{2} + {\color{red}(k+1)}$.

Think of it like a line of dominoes. The base case tips the first domino. The inductive step says "if domino $k$ falls, it knocks over domino $k+1$." In code, `sum_to(3)` waits on `sum_to(2)`, which waits on `sum_to(1)`, which waits on `sum_to(0)`. The base case answers, and then the answers fall back up the chain, one domino at a time.

This also tells us what can go wrong:

- **No base case?** Induction has no first domino, so nothing ever falls. The recursion never stops (hello, `RecursionError`).
- **Recursive call doesn't get smaller?** If we wrote `sum_to(n)` instead of `sum_to(n - 1)`, we'd never march toward the base case. The hypothesis has to be about a *smaller* problem.

So whenever you write a recursive function, ask yourself the induction questions: *What's my base case? If I trust the call on a smaller input, can I build the answer for this one?* If both answers are yes, the function is correct, and you've just proved it by induction.
